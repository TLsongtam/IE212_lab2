```python
FILL_FEATURES = ['PM2.5', 'TSP', 'O3', 'NO2', 'SO2', 'Temperature', 'Humidity']
MAX_GAP_FILL = 24  # giờ — không fill gap dài hơn 24h
 
def fill_short_gaps(series, max_gap=24):
    """
    Fill null theo thứ tự:
    1. Gap = 1h: dùng giá trị liền trước (forward fill 1 bước)
    2. Gap > 1h đến max_gap: dùng cùng giờ ngày hôm trước (lag 24h)
    Không fill gap dài hơn max_gap.
    """
    s = series.copy()
    
    # Đánh dấu vị trí null ban đầu
    null_mask = s.isnull()
    
    if not null_mask.any():
        return s, 0, 0
    
    # Tính độ dài gap tại mỗi vị trí null
    # Gán group id cho mỗi khối null liên tục
    group = (null_mask != null_mask.shift()).cumsum()
    gap_lengths = null_mask.groupby(group).transform('sum')
    
    filled_ffill = 0
    filled_lag24 = 0
    
    # Gap = 1h → forward fill
    mask_1h = null_mask & (gap_lengths == 1)
    s[mask_1h] = s.shift(1)[mask_1h]
    filled_ffill = mask_1h.sum()
    
    # Cập nhật null_mask
    null_mask = s.isnull()
    group = (null_mask != null_mask.shift()).cumsum()
    gap_lengths = null_mask.groupby(group).transform('sum')
    
    # Gap 2h đến max_gap → dùng lag 24h
    mask_lag = null_mask & (gap_lengths > 1) & (gap_lengths <= max_gap)
    lag24 = s.shift(24)
    s[mask_lag] = lag24[mask_lag]
    filled_lag24 = mask_lag.sum()
 
    return s, filled_ffill, filled_lag24
 
null_before_fill = data[FILL_FEATURES].isnull().sum().sum()
 
results = []
for station in sorted(data['Station_No'].unique()):
    idx = data[data['Station_No'] == station].index
    sub = data.loc[idx].copy()
    
    station_ffill = 0
    station_lag24 = 0
    station_unfilled = 0
    
    for col in FILL_FEATURES:
        # Bỏ qua CO trạm 2
        if station == 2 and col == 'CO':
            continue
            
        filled, n_ffill, n_lag24 = fill_short_gaps(sub[col], MAX_GAP_FILL)
        sub[col] = filled
        station_ffill += n_ffill
        station_lag24 += n_lag24
    
    station_unfilled = sub[FILL_FEATURES].isnull().sum().sum()
    data.loc[idx, FILL_FEATURES] = sub[FILL_FEATURES]
    
    print(f"  Trạm {station}: fill ffill={station_ffill}, "
          f"lag24={station_lag24}, còn null={station_unfilled}")
```

```python
INPUT_FEATURES = ['PM2.5', 'TSP', 'O3', 'CO', 'NO2', 'SO2', 'Temperature', 'Humidity']
TARGET = 'PM2.5'
 
# ============================================================
# HÀM TẠO SLIDING WINDOW CHO 1 TRẠM
# ============================================================
def create_windows(station_df, input_len, output_len, features, target):
    """
    Tạo sliding window (X, y) cho 1 trạm.
 
    Parameters:
        station_df  : DataFrame của 1 trạm, đã sort theo date
        input_len   : số giờ input  (48 hoặc 72)
        output_len  : số giờ output (24)
        features    : list feature dùng làm input
        target      : tên cột target (PM2.5)
 
    Returns:
        X : np.array shape (n_samples, input_len, n_features)
        y : np.array shape (n_samples, output_len)
        dates_X : list timestamp đầu của mỗi window
        dates_y : list timestamp cuối của mỗi window (bắt đầu output)
    """
    data = station_df[features].values          # (T, F)
    target_data = station_df[target].values     # (T,)
    dates = station_df['date'].values           # (T,)
 
    X_list, y_list = [], []
    dates_X_list, dates_y_list = [], []
 
    total = len(data)
    for i in range(total - input_len - output_len + 1):
        x_window = data[i : i + input_len]           # (input_len, F)
        y_window = target_data[i + input_len : i + input_len + output_len]  # (output_len,)
 
        # Loại window có NaN trong bất kỳ feature nào của X
        if np.isnan(x_window).any():
            continue
 
        # Loại window có NaN trong target (y)
        if np.isnan(y_window).any():
            continue
 
        X_list.append(x_window)
        y_list.append(y_window)
        dates_X_list.append(dates[i])
        dates_y_list.append(dates[i + input_len])
 
    if len(X_list) == 0:
        return None, None, None, None
 
    X = np.array(X_list, dtype=np.float32)   # (n, input_len, F)
    y = np.array(y_list, dtype=np.float32)   # (n, output_len)
 
    return X, y, dates_X_list, dates_y_list
 
 
# ============================================================
# HÀM TẠO WINDOW CHO TẤT CẢ TRẠM + TRAIN/TEST SPLIT
# ============================================================
def build_all_stations(df, input_len, output_len,
                       features=INPUT_FEATURES, target=TARGET,
                       train_ratio=0.8, val_ratio=0.1):
    """
    Tạo window cho tất cả trạm, split theo thời gian (không random).
    Tỷ lệ: train 80% / val 10% / test 10%
 
    Returns dict:
        {station_id: {'X_train', 'y_train', 'X_val', 'y_val',
                      'X_test', 'y_test', 'n_windows', 'n_features'}}
    """
    result = {}
 
    for station in sorted(df['Station_No'].unique()):
        sub = df[df['Station_No'] == station].reset_index(drop=True)

        if station == 2:
            features_used = [f for f in features if f != 'CO']
        else:
            features_used = features

        X, y, dates_X, dates_y = create_windows(
            sub, input_len, output_len, features_used, target
        )
 
        if X is None:
            print(f"  Trạm {station}: không tạo được window nào")
            continue
 
        n = len(X)
        n_train = int(n * train_ratio)
        n_val   = int(n * val_ratio)
        n_test  = n - n_train - n_val
 
        result[station] = {
            'X_train': X[:n_train],
            'y_train': y[:n_train],
            'X_val'  : X[n_train : n_train + n_val],
            'y_val'  : y[n_train : n_train + n_val],
            'X_test' : X[n_train + n_val :],
            'y_test' : y[n_train + n_val :],
            # Lưu dates để debug / inverse transform
            'dates_X_train': dates_X[:n_train],
            'dates_X_val'  : dates_X[n_train : n_train + n_val],
            'dates_X_test' : dates_X[n_train + n_val :],
            'n_windows' : n,
            'n_features': len(features),
        }
 
        print(f"  Trạm {station}: "
              f"total={n:,} | "
              f"train={n_train:,} | "
              f"val={n_val:,} | "
              f"test={n_test:,}")
 
    return result
 
 
# ============================================================
# TẠO 2 BỘ WINDOW: 48-24 VÀ 72-24
# ============================================================
print("=" * 60)
print("CẤU HÌNH 1: INPUT=48h → OUTPUT=24h")
print("=" * 60)
windows_48_24 = build_all_stations(data, input_len=48, output_len=24)
 
print("\n" + "=" * 60)
print("CẤU HÌNH 2: INPUT=72h → OUTPUT=24h")
print("=" * 60)
windows_72_24 = build_all_stations(data, input_len=72, output_len=24)
 
 
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
```

```python
# ============================================================
# 1. LOCAL ATTENTION MECHANISM (LAM)
# ============================================================
class LocalAttention(nn.Module):
    """
    Local Attention Mechanism (LAM):
    Thay vì attend toàn bộ sequence, chỉ attend trong
    cửa sổ cục bộ kích thước `window_size` quanh mỗi timestep.
    Giúp model tập trung vào các pattern ngắn hạn quan trọng.
    """
    def __init__(self, hidden_size, window_size=3):
        super().__init__()
        self.window_size = window_size
 
        # Query, Key, Value projections
        self.W_q = nn.Linear(hidden_size, hidden_size, bias=False)
        self.W_k = nn.Linear(hidden_size, hidden_size, bias=False)
        self.W_v = nn.Linear(hidden_size, hidden_size, bias=False)
 
        self.scale = hidden_size ** 0.5
        self.out_proj = nn.Linear(hidden_size, hidden_size)
 
    def forward(self, x):
        """
        x: (batch, seq_len, hidden_size)
        return: (batch, seq_len, hidden_size)
        """
        B, T, H = x.size()
        w = self.window_size
 
        Q = self.W_q(x)   # (B, T, H)
        K = self.W_k(x)   # (B, T, H)
        V = self.W_v(x)   # (B, T, H)
 
        # Pad K và V để xử lý biên
        pad = w // 2
        K_pad = F.pad(K.permute(0, 2, 1), (pad, pad), mode='replicate').permute(0, 2, 1)
        V_pad = F.pad(V.permute(0, 2, 1), (pad, pad), mode='replicate').permute(0, 2, 1)
        # K_pad, V_pad: (B, T + 2*pad, H)
 
        out = []
        for t in range(T):
            # Lấy local window quanh timestep t
            k_local = K_pad[:, t : t + w, :]    # (B, w, H)
            v_local = V_pad[:, t : t + w, :]    # (B, w, H)
            q_t     = Q[:, t, :].unsqueeze(1)   # (B, 1, H)
 
            # Attention score
            score = torch.bmm(q_t, k_local.transpose(1, 2)) / self.scale  # (B, 1, w)
            attn  = F.softmax(score, dim=-1)                               # (B, 1, w)
 
            # Context vector
            context = torch.bmm(attn, v_local)   # (B, 1, H)
            out.append(context)
 
        out = torch.cat(out, dim=1)   # (B, T, H)
        return self.out_proj(out)
 
 
# ============================================================
# 2. MULTI-SCALE BLOCK
# ============================================================
class MultiScaleBlock(nn.Module):
    """
    Multi-Scale Local Attention Module (MLAM):
    Áp dụng LocalAttention với nhiều window_size khác nhau
    để bắt pattern ở nhiều tầm thời gian cùng lúc.
    Kết quả được concat và chiếu lại về hidden_size.
    """
    def __init__(self, hidden_size, window_sizes=(3, 6, 12)):
        super().__init__()
        self.attentions = nn.ModuleList([
            LocalAttention(hidden_size, ws) for ws in window_sizes
        ])
        # Concat tất cả scales → chiếu về hidden_size
        self.fusion = nn.Linear(hidden_size * len(window_sizes), hidden_size)
        self.norm   = nn.LayerNorm(hidden_size)
 
    def forward(self, x):
        """
        x: (batch, seq_len, hidden_size)
        return: (batch, seq_len, hidden_size)
        """
        outs = [attn(x) for attn in self.attentions]  # list of (B, T, H)
        concat = torch.cat(outs, dim=-1)               # (B, T, H * n_scales)
        fused  = self.fusion(concat)                   # (B, T, H)
        return self.norm(fused + x)                    # residual connection
 
 
# ============================================================
# 3. BiLSTM-MLAM MODEL
# ============================================================
class BiLSTM_MLAM(nn.Module):
    """
    BiLSTM-MLAM: Multi-Scale Time Series Prediction Model
    dựa trên Bidirectional LSTM và Local Attention Mechanisms.
 
    Kiến trúc:
        Input → BiLSTM (layer 1) → MLAM → BiLSTM (layer 2)
              → MLAM → Dropout → FC → Output
 
    Parameters:
        input_size    : số feature đầu vào (số cột)
        hidden_size   : số unit LSTM mỗi chiều
        output_size   : số bước dự báo (24)
        num_layers    : số lớp BiLSTM (mặc định 2)
        window_sizes  : các kích thước cửa sổ local attention
        dropout       : tỷ lệ dropout
    """
    def __init__(
        self,
        input_size,
        hidden_size  = 64,
        output_size  = 24,
        num_layers   = 2,
        window_sizes = (3, 6, 12),
        dropout      = 0.2,
    ):
        super().__init__()
        self.hidden_size = hidden_size
        self.num_layers  = num_layers
 
        # BiLSTM layers xen kẽ với MLAM
        # hidden của BiLSTM = hidden_size * 2 (vì 2 chiều)
        bilstm_output = hidden_size * 2
 
        self.bilstm_layers = nn.ModuleList()
        self.mlam_layers   = nn.ModuleList()
 
        for i in range(num_layers):
            in_size = input_size if i == 0 else bilstm_output
            self.bilstm_layers.append(
                nn.LSTM(
                    input_size   = in_size,
                    hidden_size  = hidden_size,
                    num_layers   = 1,
                    batch_first  = True,
                    bidirectional= True,
                )
            )
            self.mlam_layers.append(
                MultiScaleBlock(bilstm_output, window_sizes)
            )
 
        self.dropout = nn.Dropout(dropout)
        self.fc      = nn.Linear(bilstm_output, output_size)
 
    def forward(self, x):
        """
        x: (batch, seq_len, input_size)
        return: (batch, output_size)
        """
        out = x
        for bilstm, mlam in zip(self.bilstm_layers, self.mlam_layers):
            out, _ = bilstm(out)   # (B, T, hidden*2)
            out    = mlam(out)     # (B, T, hidden*2)
 
        # Lấy timestep cuối cùng làm representation
        out = out[:, -1, :]        # (B, hidden*2)
        out = self.dropout(out)
        out = self.fc(out)         # (B, output_size)
        return out
 
 
# ============================================================
# 4. KHỞI TẠO MODEL — 2 CẤU HÌNH
# ============================================================
def build_model(input_size, config='48_24', station_id=None):
    """
    Khởi tạo model theo cấu hình input window.
 
    config: '48_24' hoặc '72_24'
    station_id: dùng để log thông tin
    """
    model = BiLSTM_MLAM(
        input_size   = input_size,
        hidden_size  = 64,
        output_size  = 24,
        num_layers   = 2,
        window_sizes = (3, 6, 12),   # multi-scale: short / mid / long
        dropout      = 0.2,
    )
 
    total_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
    label = f"Trạm {station_id}" if station_id else ""
    print(f"  [{config}] {label} | input_size={input_size} "
          f"| params={total_params:,}")
    return model
 
 
# ============================================================
# 5. SỬ DỤNG VỚI SLIDING WINDOW ĐÃ TẠO
# ============================================================
if __name__ == '__main__':
 
    # Giả sử đã có windows_48_24 và windows_72_24 từ sliding_window.py
    # from sliding_window import windows_48_24, windows_72_24
 
    # Số feature mặc định = 8, Trạm 2 = 7 (bỏ CO)
    N_FEATURES_DEFAULT = 8
    N_FEATURES_STN2    = 7
 
    print("=" * 60)
    print("KHỞI TẠO MODEL — CẤU HÌNH 48-24")
    print("=" * 60)
    models_48_24 = {}
    for station in [1, 2, 3, 4, 5, 6]:
        n_feat = N_FEATURES_STN2 if station == 2 else N_FEATURES_DEFAULT
        models_48_24[station] = build_model(n_feat, '48_24', station)
 
    print("\n" + "=" * 60)
    print("KHỞI TẠO MODEL — CẤU HÌNH 72-24")
    print("=" * 60)
    models_72_24 = {}
    for station in [1, 2, 3, 4, 5, 6]:
        n_feat = N_FEATURES_STN2 if station == 2 else N_FEATURES_DEFAULT
        models_72_24[station] = build_model(n_feat, '72_24', station)
 
    # Kiểm tra forward pass với dummy input
    print("\n" + "=" * 60)
    print("KIỂM TRA FORWARD PASS")
    print("=" * 60)
    dummy_48 = torch.randn(32, 48, 8)   # batch=32, seq=48, features=8
    dummy_72 = torch.randn(32, 72, 8)
 
    out_48 = models_48_24[1](dummy_48)
    out_72 = models_72_24[1](dummy_72)
 
    print(f"  Input 48h  → Output: {out_48.shape}")   # (32, 24)
    print(f"  Input 72h  → Output: {out_72.shape}")   # (32, 24)
    print("\n✅ Model sẵn sàng.")
```

```python
import torch
import torch.nn as nn
import numpy as np
from torch.utils.data import DataLoader, TensorDataset
from sklearn.preprocessing import MinMaxScaler
```

```python
def get_scaler(X_train, y_train):
    """
    Fit MinMaxScaler trên train set.
    X: (n, seq_len, n_features) → reshape về 2D để fit
    y: (n, 24)
    Trả về scaler_X, scaler_y đã fit.
    """
    n, T, F = X_train.shape
    scaler_X = MinMaxScaler()
    scaler_y = MinMaxScaler()
 
    scaler_X.fit(X_train.reshape(-1, F))
    scaler_y.fit(y_train)
 
    return scaler_X, scaler_y
 
 
def scale_data(X, y, scaler_X, scaler_y):
    """Transform X và y theo scaler đã fit."""
    n, T, F = X.shape
    X_scaled = scaler_X.transform(X.reshape(-1, F)).reshape(n, T, F)
    y_scaled = scaler_y.transform(y)
    return X_scaled.astype(np.float32), y_scaled.astype(np.float32)
 
 
def make_loader(X, y, batch_size, shuffle=True):
    """Tạo DataLoader từ numpy array."""
    dataset = TensorDataset(
        torch.tensor(X), 
        torch.tensor(y)
    )
    return DataLoader(dataset, batch_size=batch_size, shuffle=shuffle)
 
 
def train_one_station(
    station_id,
    model,
    windows,
    device,
    epochs     = 100,
    batch_size = 32,
    lr         = 1e-3,
    patience   = 15,       # early stopping
    config     = '',
):
    """
    Train 1 model cho 1 trạm.
 
    Returns:
        model       : model đã train (best weights)
        scalers     : (scaler_X, scaler_y) để inverse transform khi evaluate
        history     : dict chứa train_loss và val_loss theo epoch
    """
    # --- Lấy data ---
    d = windows[station_id]
    X_train, y_train = d['X_train'], d['y_train']
    X_val,   y_val   = d['X_val'],   d['y_val']
 
    # --- Scale ---
    scaler_X, scaler_y = get_scaler(X_train, y_train)
    X_tr, y_tr = scale_data(X_train, y_train, scaler_X, scaler_y)
    X_vl, y_vl = scale_data(X_val,   y_val,   scaler_X, scaler_y)
 
    # --- DataLoader ---
    train_loader = make_loader(X_tr, y_tr, batch_size, shuffle=True)
    val_loader   = make_loader(X_vl, y_vl, batch_size, shuffle=False)
 
    # --- Optimizer & Loss ---
    optimizer = torch.optim.Adam(model.parameters(), lr=lr)
    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer, mode='min', factor=0.5, patience=7
    )
    criterion = nn.MSELoss()
 
    model.to(device)
 
    # --- Training loop ---
    best_val_loss  = float('inf')
    best_weights   = None
    no_improve     = 0
    history        = {'train_loss': [], 'val_loss': []}
 
    for epoch in range(1, epochs + 1):
 
        # Train
        model.train()
        train_losses = []
        for X_batch, y_batch in train_loader:
            X_batch = X_batch.to(device)
            y_batch = y_batch.to(device)
 
            optimizer.zero_grad()
            pred = model(X_batch)
            loss = criterion(pred, y_batch)
            loss.backward()
            nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
            optimizer.step()
            train_losses.append(loss.item())
 
        # Validation
        model.eval()
        val_losses = []
        with torch.no_grad():
            for X_batch, y_batch in val_loader:
                X_batch = X_batch.to(device)
                y_batch = y_batch.to(device)
                pred = model(X_batch)
                loss = criterion(pred, y_batch)
                val_losses.append(loss.item())
 
        train_loss = np.mean(train_losses)
        val_loss   = np.mean(val_losses)
        history['train_loss'].append(train_loss)
        history['val_loss'].append(val_loss)
 
        scheduler.step(val_loss)
 
        # Early stopping
        if val_loss < best_val_loss:
            best_val_loss = val_loss
            best_weights  = {k: v.cpu().clone() for k, v in model.state_dict().items()}
            no_improve    = 0
        else:
            no_improve += 1
            if no_improve >= patience:
                print(f"    Early stop tại epoch {epoch} "
                      f"(best val_loss={best_val_loss:.6f})")
                break
 
        if epoch % 10 == 0:
            print(f"    Epoch {epoch:3d} | "
                  f"train={train_loss:.4f} | "
                  f"val={val_loss:.4f} | "
                  f"best={best_val_loss:.4f}")
 
    # Load best weights
    model.load_state_dict(best_weights)
    model.eval()
 
    # Lưu model
    save_path = f"model_{config}_station{station_id}.pt"
    torch.save({
        'model_state_dict': best_weights,
        'scaler_X'        : scaler_X,
        'scaler_y'        : scaler_y,
        'history'         : history,
        'config'          : config,
        'station_id'      : station_id,
    }, save_path)
    print(f"    ✅ Saved: {save_path}")
 
    return model, (scaler_X, scaler_y), history
 
 
# ============================================================
# SETUP CHUNG
# ============================================================
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Device: {device}")
 
EPOCHS     = 100
BATCH_SIZE = 32
LR         = 1e-3
PATIENCE   = 15
```

```python
import torch
import numpy as np
from sklearn.metrics import mean_squared_error, mean_absolute_error
from torch.utils.data import DataLoader, TensorDataset
 
 
# ============================================================
# HÀM TÍNH METRIC
# ============================================================
def rmse(y_true, y_pred):
    return np.sqrt(mean_squared_error(y_true, y_pred))
 
def mae(y_true, y_pred):
    return mean_absolute_error(y_true, y_pred)
 
def mape(y_true, y_pred, eps=1e-8):
    return np.mean(np.abs((y_true - y_pred) / (np.abs(y_true) + eps))) * 100
 
 
# ============================================================
# HÀM PREDICT TRÊN TEST SET
# ============================================================
def predict_test(model, windows, station_id, scalers, device, batch_size=32):
    """
    Chạy inference trên test set của 1 trạm.
    Inverse transform về đơn vị gốc (μg/m³) trước khi tính metric.
 
    Returns:
        y_true : (n_test, 24) — giá trị thực, đơn vị gốc
        y_pred : (n_test, 24) — giá trị dự báo, đơn vị gốc
    """
    scaler_X, scaler_y = scalers
    d = windows[station_id]
 
    X_test = d['X_test']   # (n, seq_len, F)
    y_test = d['y_test']   # (n, 24)
 
    # Scale X_test
    n, T, F = X_test.shape
    X_scaled = scaler_X.transform(X_test.reshape(-1, F)).reshape(n, T, F)
    X_scaled = X_scaled.astype(np.float32)
 
    # DataLoader
    loader = DataLoader(
        TensorDataset(torch.tensor(X_scaled)),
        batch_size=batch_size,
        shuffle=False
    )
 
    # Inference
    model.eval()
    preds = []
    with torch.no_grad():
        for (X_batch,) in loader:
            X_batch = X_batch.to(device)
            out = model(X_batch)           # (B, 24) — scaled
            preds.append(out.cpu().numpy())
 
    y_pred_scaled = np.concatenate(preds, axis=0)   # (n, 24)
 
    # Inverse transform về μg/m³
    y_pred = scaler_y.inverse_transform(y_pred_scaled)
    y_true = y_test   # y_test đã ở đơn vị gốc, không cần inverse
 
    return y_true, y_pred
 
 
# ============================================================
# HÀM TÍNH METRIC CHO 1 TRẠM
# ============================================================
def evaluate_station(model, windows, station_id, scalers, device,
                     config='', verbose=True):
    """
    Tính RMSE, MAE, MAPE trên toàn bộ 24 bước dự báo
    và trung bình qua tất cả test samples.
    """
    y_true, y_pred = predict_test(
        model, windows, station_id, scalers, device
    )
 
    # Flatten (n, 24) → (n*24,) để tính metric tổng thể
    y_true_flat = y_true.flatten()
    y_pred_flat = y_pred.flatten()
 
    metrics = {
        'RMSE': rmse(y_true_flat, y_pred_flat),
        'MAE' : mae(y_true_flat, y_pred_flat),
        'MAPE': mape(y_true_flat, y_pred_flat),
    }
 
    if verbose:
        print(f"  [{config}] Trạm {station_id} | "
              f"RMSE={metrics['RMSE']:.4f} | "
              f"MAE={metrics['MAE']:.4f} | "
              f"MAPE={metrics['MAPE']:.2f}%")
 
    return metrics, y_true, y_pred
 
 
# ============================================================
# HÀM ĐÁNH GIÁ TẤT CẢ TRẠM — 1 CẤU HÌNH
# ============================================================
def evaluate_all_stations(trained, windows, config, device):
    """
    Chạy evaluate cho toàn bộ 6 trạm, in bảng tổng kết.
 
    Parameters:
        trained : dict {station_id: {'model', 'scalers', 'history'}}
        windows : windows_48_24 hoặc windows_72_24
        config  : '48_24' hoặc '72_24'
        device  : torch.device
 
    Returns:
        all_metrics : dict {station_id: {'RMSE', 'MAE', 'MAPE'}}
        all_preds   : dict {station_id: (y_true, y_pred)}
    """
    print(f"\n{'='*60}")
    print(f"EVALUATE — CẤU HÌNH {config}")
    print(f"{'='*60}")
 
    all_metrics = {}
    all_preds   = {}
 
    for station in sorted(trained.keys()):
        model   = trained[station]['model']
        scalers = trained[station]['scalers']
 
        metrics, y_true, y_pred = evaluate_station(
            model, windows, station, scalers, device,
            config=config, verbose=True
        )
        all_metrics[station] = metrics
        all_preds[station]   = (y_true, y_pred)
 
    # Bảng tổng kết
    print(f"\n{'─'*50}")
    print(f"  {'Trạm':>6} | {'RMSE':>8} | {'MAE':>8} | {'MAPE':>8}")
    print(f"{'─'*50}")
    rmse_list, mae_list, mape_list = [], [], []
    for s, m in all_metrics.items():
        print(f"  {s:>6} | {m['RMSE']:>8.4f} | {m['MAE']:>8.4f} | {m['MAPE']:>7.2f}%")
        rmse_list.append(m['RMSE'])
        mae_list.append(m['MAE'])
        mape_list.append(m['MAPE'])
    print(f"{'─'*50}")
    print(f"  {'Mean':>6} | {np.mean(rmse_list):>8.4f} | "
          f"{np.mean(mae_list):>8.4f} | {np.mean(mape_list):>7.2f}%")
    print(f"{'─'*50}")
 
    return all_metrics, all_preds
 
 
# ============================================================
# CELL EVALUATE — GỌI SAU KHI TRAIN XONG
# ============================================================
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
 
# Evaluate cấu hình 48-24
metrics_48_24, preds_48_24 = evaluate_all_stations(
    trained = trained_48_24,
    windows = windows_48_24,
    config  = '48_24',
    device  = device,
)
 
# Evaluate cấu hình 72-24
metrics_72_24, preds_72_24 = evaluate_all_stations(
    trained = trained_72_24,
    windows = windows_72_24,
    config  = '72_24',
    device  = device,
)
 
# So sánh 2 cấu hình
print(f"\n{'='*60}")
print("SO SÁNH 48-24 vs 72-24 (trung bình 6 trạm)")
print(f"{'='*60}")
for metric in ['RMSE', 'MAE', 'MAPE']:
    val_48 = np.mean([metrics_48_24[s][metric] for s in metrics_48_24])
    val_72 = np.mean([metrics_72_24[s][metric] for s in metrics_72_24])
    better = '48h ✅' if val_48 < val_72 else '72h ✅'
    unit   = '%' if metric == 'MAPE' else 'μg/m³'
    print(f"  {metric:>4}: 48h={val_48:.4f}{unit} | "
          f"72h={val_72:.4f}{unit} | winner={better}")
```
