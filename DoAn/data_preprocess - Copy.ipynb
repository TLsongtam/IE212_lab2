```python
import numpy as np

valid_range = {
    'PM2.5':       (0, 500),      # μg/m³
    'TSP':         (0, 1000),     # μg/m³
    'O3':          (0, 500),      # μg/m³
    'CO':          (0, 15000),    # μg/m³ — mean ~994, loại spike > 10000
    'NO2':         (0, 500),      # μg/m³
    'SO2':         (0, 700),      # μg/m³ — max thực tế ~700
    'Temperature': (-10, 50),     # °C
    'Humidity':    (0, 100),      # %
}


 
for col, (lo, hi) in valid_range.items():
    mask = (data[col] < lo) | (data[col] > hi)
    n = mask.sum()
    if n > 0:
        print(f"  [{col}] Loại {n} giá trị nằm ngoài [{lo}, {hi}]")
        data.loc[mask, col] = np.nan
 

```

```python
import numpy as np
import pandas as pd

outlier_summary = []
total_outliers = 0

for station in sorted(data['Station_No'].unique()):
    idx = data[data['Station_No'] == station].index
    pm = data.loc[idx, 'PM2.5'].copy()
    pm_prev = pm.shift(1)
 
    outlier_mask = (pm > 200) & (pm > 3 * pm_prev)
    n = outlier_mask.sum()
    
    if n > 0:
        total_outliers += n
       
        outlier_summary.append({
            'Station_No': station, 
            'Outlier_Count': n
        })


    data.loc[idx[outlier_mask], 'PM2.5'] = np.nan


df_outliers = pd.DataFrame(outlier_summary)


print(f"Tổng số outlier: {total_outliers}")
print(df_outliers)
```

```python
FILL_FEATURES = ['PM2.5', 'TSP', 'O3', 'NO2', 'SO2', 'Temperature', 'Humidity']
MAX_GAP_FILL = 24  
 
def fill_short_gaps(series, max_gap=24):

    s = series.copy()
    
    # Đánh dấu vị trí null ban đầu
    null_mask = s.isnull()
    
    if not null_mask.any():
        return s, 0, 0
    

    group = (null_mask != null_mask.shift()).cumsum()
    gap_lengths = null_mask.groupby(group).transform('sum')
    
    filled_ffill = 0
    filled_lag24 = 0
    

    mask_1h = null_mask & (gap_lengths == 1)
    s[mask_1h] = s.shift(1)[mask_1h]
    filled_ffill = mask_1h.sum()
    
  
    null_mask = s.isnull()
    group = (null_mask != null_mask.shift()).cumsum()
    gap_lengths = null_mask.groupby(group).transform('sum')
    

    mask_lag = null_mask & (gap_lengths > 1) & (gap_lengths <= max_gap)
    lag24 = s.shift(24)
    s[mask_lag] = lag24[mask_lag]
    filled_lag24 = mask_lag.sum()
 
    return s, filled_ffill, filled_lag24
 
null_before_fill = data[FILL_FEATURES].isnull().sum().sum()
 
results = []
for station in sorted(data['Station_No'].unique()):
    idx = data[data['Station_No'] == station].index
    sub = data.loc[idx].copy()
    
    station_ffill = 0
    station_lag24 = 0
    station_unfilled = 0
    
    for col in FILL_FEATURES:

        if station == 2 and col == 'CO':
            continue
            
        filled, n_ffill, n_lag24 = fill_short_gaps(sub[col], MAX_GAP_FILL)
        sub[col] = filled
        station_ffill += n_ffill
        station_lag24 += n_lag24
    
    station_unfilled = sub[FILL_FEATURES].isnull().sum().sum()
    data.loc[idx, FILL_FEATURES] = sub[FILL_FEATURES]
    
    print(f"  Trạm {station}: fill ffill={station_ffill}, "
          f"lag24={station_lag24}, còn null={station_unfilled}")
```

```python
INPUT_FEATURES = ['PM2.5', 'TSP', 'O3', 'CO', 'NO2', 'SO2', 'Temperature', 'Humidity']
TARGET = 'PM2.5'
 

def create_windows(station_df, input_len, output_len, features, target):
    
    data = station_df[features].values          # (T, F)
    target_data = station_df[target].values     # (T,)
    dates = station_df['date'].values           # (T,)
 
    X_list, y_list = [], []
    dates_X_list, dates_y_list = [], []
 
    total = len(data)
    for i in range(total - input_len - output_len + 1):
        x_window = data[i : i + input_len]           # (input_len, F)
        y_window = target_data[i + input_len : i + input_len + output_len]  # (output_len,)
 
        
        if np.isnan(x_window).any():
            continue
 
        
        if np.isnan(y_window).any():
            continue
 
        X_list.append(x_window)
        y_list.append(y_window)
        dates_X_list.append(dates[i])
        dates_y_list.append(dates[i + input_len])
 
    if len(X_list) == 0:
        return None, None, None, None
 
    X = np.array(X_list, dtype=np.float32)   
    y = np.array(y_list, dtype=np.float32)   
 
    return X, y, dates_X_list, dates_y_list
 
 

def build_all_stations(df, input_len, output_len,
                       features=INPUT_FEATURES, target=TARGET,
                       train_ratio=0.8, val_ratio=0.1):
    
    result = {}
 
    for station in sorted(df['Station_No'].unique()):
        sub = df[df['Station_No'] == station].reset_index(drop=True)

        if station == 2:
            features_used = [f for f in features if f != 'CO']
        else:
            features_used = features

        X, y, dates_X, dates_y = create_windows(
            sub, input_len, output_len, features_used, target
        )
 
        if X is None:
            print(f"  Trạm {station}: không tạo được window nào")
            continue
 
        n = len(X)
        n_train = int(n * train_ratio)
        n_val   = int(n * val_ratio)
        n_test  = n - n_train - n_val
 
        result[station] = {
            'X_train': X[:n_train],
            'y_train': y[:n_train],
            'X_val'  : X[n_train : n_train + n_val],
            'y_val'  : y[n_train : n_train + n_val],
            'X_test' : X[n_train + n_val :],
            'y_test' : y[n_train + n_val :],
            
            'dates_X_train': dates_X[:n_train],
            'dates_X_val'  : dates_X[n_train : n_train + n_val],
            'dates_X_test' : dates_X[n_train + n_val :],
            'n_windows' : n,
            'n_features': len(features),
        }
 
        print(f"  Trạm {station}: "
              f"total={n:,} | "
              f"train={n_train:,} | "
              f"val={n_val:,} | "
              f"test={n_test:,}")
 
    return result
 
 

print("=" * 60)
print("CẤU HÌNH 1: INPUT=48h → OUTPUT=24h")
print("=" * 60)
windows_48_24 = build_all_stations(data, input_len=48, output_len=24)
 
print("\n" + "=" * 60)
print("CẤU HÌNH 2: INPUT=72h → OUTPUT=24h")
print("=" * 60)
windows_72_24 = build_all_stations(data, input_len=72, output_len=24)
 
 
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
```

```python

class LocalAttention(nn.Module):
    """
    Local Attention Mechanism (LAM):
    Thay vì attend toàn bộ sequence, chỉ attend trong
    cửa sổ cục bộ kích thước `window_size` quanh mỗi timestep.
    Giúp model tập trung vào các pattern ngắn hạn quan trọng.
    """
    def __init__(self, hidden_size, window_size=3):
        super().__init__()
        self.window_size = window_size
 
        # Query, Key, Value projections
        self.W_q = nn.Linear(hidden_size, hidden_size, bias=False)
        self.W_k = nn.Linear(hidden_size, hidden_size, bias=False)
        self.W_v = nn.Linear(hidden_size, hidden_size, bias=False)
 
        self.scale = hidden_size ** 0.5
        self.out_proj = nn.Linear(hidden_size, hidden_size)
 
    def forward(self, x):
        """
        x: (batch, seq_len, hidden_size)
        return: (batch, seq_len, hidden_size)
        """
        B, T, H = x.size()
        w = self.window_size
 
        Q = self.W_q(x)   # (B, T, H)
        K = self.W_k(x)   # (B, T, H)
        V = self.W_v(x)   # (B, T, H)
 
        # Pad K và V để xử lý biên
        pad = w // 2
        K_pad = F.pad(K.permute(0, 2, 1), (pad, pad), mode='replicate').permute(0, 2, 1)
        V_pad = F.pad(V.permute(0, 2, 1), (pad, pad), mode='replicate').permute(0, 2, 1)
        # K_pad, V_pad: (B, T + 2*pad, H)
 
        out = []
        for t in range(T):
            # Lấy local window quanh timestep t
            k_local = K_pad[:, t : t + w, :]    # (B, w, H)
            v_local = V_pad[:, t : t + w, :]    # (B, w, H)
            q_t     = Q[:, t, :].unsqueeze(1)   # (B, 1, H)
 
            # Attention score
            score = torch.bmm(q_t, k_local.transpose(1, 2)) / self.scale  # (B, 1, w)
            attn  = F.softmax(score, dim=-1)                               # (B, 1, w)
 
            # Context vector
            context = torch.bmm(attn, v_local)   # (B, 1, H)
            out.append(context)
 
        out = torch.cat(out, dim=1)   # (B, T, H)
        return self.out_proj(out)
 
 

class MultiScaleBlock(nn.Module):
    """
    Multi-Scale Local Attention Module (MLAM):
    Áp dụng LocalAttention với nhiều window_size khác nhau
    để bắt pattern ở nhiều tầm thời gian cùng lúc.
    Kết quả được concat và chiếu lại về hidden_size.
    """
    def __init__(self, hidden_size, window_sizes=(3, 6, 12)):
        super().__init__()
        self.attentions = nn.ModuleList([
            LocalAttention(hidden_size, ws) for ws in window_sizes
        ])
        # Concat tất cả scales → chiếu về hidden_size
        self.fusion = nn.Linear(hidden_size * len(window_sizes), hidden_size)
        self.norm   = nn.LayerNorm(hidden_size)
 
    def forward(self, x):
        """
        x: (batch, seq_len, hidden_size)
        return: (batch, seq_len, hidden_size)
        """
        outs = [attn(x) for attn in self.attentions]  # list of (B, T, H)
        concat = torch.cat(outs, dim=-1)               # (B, T, H * n_scales)
        fused  = self.fusion(concat)                   # (B, T, H)
        return self.norm(fused + x)                    # residual connection
 
 

class BiLSTM_MLAM(nn.Module):

    def __init__(
        self,
        input_size,
        hidden_size  = 64,
        output_size  = 24,
        num_layers   = 2,
        window_sizes = (3, 6, 12),
        dropout      = 0.2,
    ):
        super().__init__()
        self.hidden_size = hidden_size
        self.num_layers  = num_layers
 

        bilstm_output = hidden_size * 2
 
        self.bilstm_layers = nn.ModuleList()
        self.mlam_layers   = nn.ModuleList()
 
        for i in range(num_layers):
            in_size = input_size if i == 0 else bilstm_output
            self.bilstm_layers.append(
                nn.LSTM(
                    input_size   = in_size,
                    hidden_size  = hidden_size,
                    num_layers   = 1,
                    batch_first  = True,
                    bidirectional= True,
                )
            )
            self.mlam_layers.append(
                MultiScaleBlock(bilstm_output, window_sizes)
            )
 
        self.dropout = nn.Dropout(dropout)
        self.fc      = nn.Linear(bilstm_output, output_size)
 
    def forward(self, x):
        """
        x: (batch, seq_len, input_size)
        return: (batch, output_size)
        """
        out = x
        for bilstm, mlam in zip(self.bilstm_layers, self.mlam_layers):
            out, _ = bilstm(out)   
            out    = mlam(out)     
 

        out = out[:, -1, :]        
        out = self.dropout(out)
        out = self.fc(out)         
        return out
 
 

def build_model(input_size, config='48_24', station_id=None):
    """
    Khởi tạo model theo cấu hình input window.
 
    config: '48_24' hoặc '72_24'
    station_id: dùng để log thông tin
    """
    model = BiLSTM_MLAM(
        input_size   = input_size,
        hidden_size  = 64,
        output_size  = 24,
        num_layers   = 2,
        window_sizes = (3, 6, 12),   # multi-scale: short / mid / long
        dropout      = 0.2,
    )
 
    total_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
    label = f"Trạm {station_id}" if station_id else ""
    print(f"  [{config}] {label} | input_size={input_size} "
          f"| params={total_params:,}")
    return model
 
 

if __name__ == '__main__':
 
    
    N_FEATURES_DEFAULT = 8
    N_FEATURES_STN2    = 7
 
    print("=" * 60)
    print("KHỞI TẠO MODEL — CẤU HÌNH 48-24")
    print("=" * 60)
    models_48_24 = {}
    for station in [1, 2, 3, 4, 5, 6]:
        n_feat = N_FEATURES_STN2 if station == 2 else N_FEATURES_DEFAULT
        models_48_24[station] = build_model(n_feat, '48_24', station)
 
    print("\n" + "=" * 60)
    print("KHỞI TẠO MODEL — CẤU HÌNH 72-24")
    print("=" * 60)
    models_72_24 = {}
    for station in [1, 2, 3, 4, 5, 6]:
        n_feat = N_FEATURES_STN2 if station == 2 else N_FEATURES_DEFAULT
        models_72_24[station] = build_model(n_feat, '72_24', station)
 
   
    print("\n" + "=" * 60)
    print("KIỂM TRA FORWARD PASS")
    print("=" * 60)
    dummy_48 = torch.randn(32, 48, 8)   
    dummy_72 = torch.randn(32, 72, 8)
 
    out_48 = models_48_24[1](dummy_48)
    out_72 = models_72_24[1](dummy_72)
 
    print(f"  Input 48h  → Output: {out_48.shape}")   
    print(f"  Input 72h  → Output: {out_72.shape}")   
    print("\n✅ Model sẵn sàng.")
```
